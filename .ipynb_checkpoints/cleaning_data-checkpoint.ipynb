```python
import pandas as pd
```

```python
# cleaning data 
```

```python
#handle missing values

df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
# they can show the null value (null value = true)
df.isnull()
```

```python
# count the missing value
df.isnull().sum()
```

```python
#drop the null values rows
df.dropna()
```

```python
#fill any number or value with the place of all null value
df.fillna(0)
```

```python
#fill the mean value of the place of age null value
age_mean = df["age"].mean()
df["age"].fillna(age_mean)
```

```python
#Now fill our data with a copy
clean_data = df.copy()
age_mean = clean_data["age"].mean()
clean_data["age"] = clean_data["age"].fillna(age_mean)
clean_data
```

```python
#forword fill means copy the data with the help of upper values
df.ffill()
```

```python
#Backword fill means copy the data with the help of down values
df.bfill()
```

```python
# handle duplicate values 
df.duplicated()
```

```python
#check one specific columns Duplicate
df["country"].duplicated()
```

```python
#drop the complete duplicate data.
df.drop_duplicates()
```

```python
#handle data type and date and time 
df.dtypes
```

```python
df2 = df.copy()
df2 = df2.fillna(0)
df2 = df2["age"].astype("int64").copy()
df2.dtypes
```
