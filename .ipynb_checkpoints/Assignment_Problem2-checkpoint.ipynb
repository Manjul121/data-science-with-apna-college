```python
salary = float(input ("Enter your salary:"))
if (salary < 30000):
    tax = 5
elif (salary <= 70000):
    tax = 15
else :
    tax = 25
Total_tax = salary*tax/100
print("Tax rate:",tax,"%")
print("tax Amount :",Total_tax)
```

```python
#only for positive numbers not for zero or negitive numbers.
def count_digits (n):
    count = 0
    while n > 0:
        count = count + 1
        n = n // 10 
    return count 
num = int (input ("Enter a Number :"))
result = count_digits (num)
print("Total digits = ",result)
```
