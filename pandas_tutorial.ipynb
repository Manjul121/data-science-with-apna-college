```python
s = pd.Series([11,12,13,19,22] , index = ["Carl","Tom","Bittu","Riya","Manjul"])
print(s)
print(s["Riya"])
print(s["Manjul"])
print(s.iloc[0])
print(s.index)

```

```python
#sum of the 2 series.
s1 =pd.Series([11,12,13,19,22])
s2 = pd.Series([1,1,1,1,1])
print(s1+s2)
s1[0] = 50
print(s1+s2)
```
