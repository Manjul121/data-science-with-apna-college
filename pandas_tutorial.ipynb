```python
#filtering data
#filtering data single line 
df[df["aqi"]>100]
```

```python
#filtering data to find a specific things
df[df["no2"]<6] 
```

```python
#filtering data to some logics
df[(df["no2"]<6) &(df["temperature"]<30)]
```

```python
#filtering data to access the or operator
df[(df["no2"]<6) |(df["temperature"]<30)]
```

```python
#filtering data with high logics
df[(df["no2"]<6) &(df["temperature"]<30) & (df["pm25"]>5)]
```

```python
# I want only city and aqi deatails of no2 <6 , temperature< 30, pm25>5 only lets filter data 
df[(df["no2"]<6) &(df["temperature"]<30) & (df["pm25"]>5)] [["city", "aqi"]]
```

```python
# filter data using Query
df.query("aqi<50 & temperature >30")[["city", "aqi"]]
```

```python
# filter data using Query and python var
aqi_value = 50
df.query("aqi< @aqi_value & temperature >30")[["city", "aqi"]]
```
